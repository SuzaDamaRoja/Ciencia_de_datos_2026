##CONDICIONALES
Evalúan una condición de verdad. Si la condición es verdadera, pasa algo, SINO, pasa otra cosa.

###if-else
sintaxis
```
if (condicion):
  pasan acciones si la condición es verdadera
else:
  pasan acciones si la condición es falsa
```



````python
pregunta=input("Quieres helado?")
if pregunta=="sí":
  print("Procedo a comprar helado 🍧")
else:
  print("te quedaste sin helado unicorni@ 🦄")
````

#IF-ELIF
Se emplea cuando hay múltiples condiciones y cualquiera puede ser verdadera. Es el reemplazo moderno del switch

Sintaxis
```
if condicion1: Ejecuta si condicion1 es verdadera
elif condicion2: Ejecuta si condicion1 es verdadera
elif condicionn: Ejecuta si condicionn es verdadera
else: ninguna de las anteriores
```



````python
sabor=input("Qué sabor quieres? ")
if sabor=="chocolate": print("Procedo a comprar helado 🍨")
elif sabor=="vainilla":print("Procedo a comprar helado 🍨")
elif sabor=="arequipe":print("Procedo a comprar helado 🍨")
elif sabor=="frutos del bosque": print("Procedo a comprar helado 🍨")
else: print("Lo siento. No hay ese sabor")
````
